```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Health data1 - Sheet1.csv',encoding='latin-1')
```

```python
expense_data = df.groupby(['Expense Area','Supplier'])
```

```python
expense_data.first()
```

```python
expense_df = df['Expense Area'].value_counts()
```

```python
expense_df.plot(kind='barh',xlabel='Expense Areas',ylabel='Frequency',title='Expense Departments')
```

Operations has the most popular frequency than other expense departments. The second biggest department was Digital, Data & Technology.

```python
supplier_data = df.groupby(['Expense Area','Supplier']).agg(Number_of_Suppliers=('Supplier','count'))
```

```python
supplier_data
```

Most popular supplier of the Operations Area was Royal Mail. Some of these supplier from the different expenses area has the least number.

```python
expense_type_data = df.groupby(['Expense Area','Expense Type']).agg(Number_of_types=('Expense Type','count'))
```
