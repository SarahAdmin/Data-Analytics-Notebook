```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
races = pd.read_csv('ethnic_data.csv')
```

```python
races.info()
```

```python
sns.barplot(data=races,x='Year',y='Trips per person per year',palette='Set1',errorbar=None)
plt.title('Trips per Person')
plt.ylabel('Average number of trips')
plt.show()
```

The bar graph shows the average number of trips per year. 2023 was the largest frequency of average number of trips. 2020 was the least popular frequency due COVID-19 pandemic.

```python
sns.countplot(data=races,y='Ethnic group',palette='Set1',orient='h')
plt.title('Ethnic Group Distribution')
plt.show()
```

Five Ethnic Groups has same frequencies of average number of trips in the United Kingdom. All Ethnic groups has lowest frequency.
